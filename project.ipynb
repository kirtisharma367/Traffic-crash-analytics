```python
#WEATHER CONDITION
query = """
SELECT WEATHER_CONDITION, COUNT(*) AS crash_count
FROM crashes
GROUP BY WEATHER_CONDITION
ORDER BY crash_count DESC
"""

weather = pd.read_sql(query, con=engine)

print(weather)
```

```python
#LIGHTING COUNTS
query = """
SELECT LIGHTING_CONDITION, COUNT(*) AS crash_count
FROM crashes
GROUP BY LIGHTING_CONDITION
ORDER BY crash_count DESC
"""

lighting = pd.read_sql(query, con=engine)

print(lighting)
```

```python
#Common Causes (Traffic Control Device)
query = """
SELECT TRAFFIC_CONTROL_DEVICE, COUNT(*) AS crash_count
FROM crashes
GROUP BY TRAFFIC_CONTROL_DEVICE
ORDER BY crash_count DESC
"""

causes = pd.read_sql(query, con=engine)

print(causes.head(10))
```

```python
#Crash Type Analysis
query = """
SELECT FIRST_CRASH_TYPE, COUNT(*) AS crash_count
FROM crashes
GROUP BY FIRST_CRASH_TYPE
ORDER BY crash_count DESC
"""

types = pd.read_sql(query, con=engine)

print(types.head(10))
```

```python
#⏰Monthly Trend

query = """
SELECT
    DATE_FORMAT(CRASH_DATE, '%%Y-%%m') AS month,
    COUNT(*) AS total_crashes
FROM crashes
GROUP BY month
ORDER BY month
"""
month=pd.read_sql(query,con=engine)

print(month.head(10))
```

```python
#📅Yearly Trend
query = """
SELECT YEAR(CRASH_DATE) AS year,
       COUNT(*) AS crash_count
FROM crashes
GROUP BY YEAR(CRASH_DATE)
ORDER BY year
"""

year = pd.read_sql(query, con=engine)
print(year)
```

```python
# Top 5 streets with highest crashes
top_streets = traffic_data.groupby('STREET_NAME').size() \
                .sort_values(ascending=False) \
                .head(5)

plt.figure(figsize=(8,5))
top_streets.plot(kind='bar')
plt.title('Top 5 High-Risk Streets')
plt.xlabel('Street Name')
plt.ylabel('Number of Crashes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Top Factors Contributing to High-Severity Crashes
severity_factors = traffic_data.groupby('WEATHER_CONDITION')['INJURIES_TOTAL'].sum() \
                     .sort_values(ascending=False) \
                     .head(5)

plt.figure(figsize=(8,5))
severity_factors.plot(kind='bar')
plt.title('Factors Contributing to High-Severity Crashes')
plt.xlabel('Weather Condition')
plt.ylabel('Total Injuries')
plt.tight_layout()
plt.show()
```

```python
#Average Number of Crashes by Hour of Day
hourly_crashes = traffic_data.groupby('CRASH_HOUR').size()

plt.figure(figsize=(10,5))
hourly_crashes.plot(kind='line', marker='o')
plt.title('Crashes by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Number of Crashes')
plt.grid(True)
plt.show()
```
